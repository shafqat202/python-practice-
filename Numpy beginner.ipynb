# 🧠 NumPy Basics: Creating & Randomizing

### Core Concepts:
*   **`np.arange(start, stop, step)`**: Works like the Python `range()` but creates a NumPy array.
*   **`np.linspace(start, stop, count)`**: Generates exactly `count` numbers evenly spaced between the start and stop.
*   **Randomness**: `np.random` generates numbers. Remember that `randint` produces whole numbers, while `random` produces decimals (floats).

```python
import numpy as np

# Creating arrays with specific patterns
arrange=np.arange(1,11,3) # Start at 1, go up to 10, jump by 3
scale=np.linspace(1,10,3) # Give me 3 numbers perfectly spaced between 1 and 10
fourdmatrix=np.arange(24).reshape(2,2,2,3) # 24 numbers packed into a 4D structure

# Random number generation
a=np.random.random(12) # 12 random decimals between 0 and 1
print(a)
print()

b=np.random.randint(12) # A single random whole number up to 12
print(b)
print()

c=np.random.randint(1,100,5) # 5 random whole numbers between 1 and 100
print(c)
print()

d=np.random.randint(1,100,size=(5,3)) # A 5x3 grid of random numbers
print(d)

# Note: To reshape random ints, you MUST provide the size inside randint()
#t=np.random.randint(1,100, size=(3,3))
#print(t)
```

# 🏗️ Structural Changes & Shapes

### Core Concepts:
*   **`.reshape(rows, cols)`**: Changes the view of the data without changing the data itself. The total size (rows * cols) must match the number of elements.
*   **`np.zeros` & `np.ones`**: Quick way to initialize empty grids for math operations.
*   **`np.identity`**: Creates a square matrix with 1s on the diagonal—very important in linear algebra!

```python
x=np.array([1,2,3,4,5,6,7,8])
# Turning a flat list into a 3D block (2x2x2)
matrix=x.reshape(2,2,2)
print(matrix)

# Quick ways to make grids filled with specific numbers
print(np.zeros((3,10),dtype=int)) # A 3x10 grid of zeros
print()

print(np.ones((3,10),dtype=int)*3) # Start with ones, then multiply everything by 3

# Identity matrix: 1s down the diagonal, 0s everywhere else
print(np.identity(4))
```

# 🔍 Metadata & Seeds

### Core Concepts:
*   **`np.random.seed()`**: This 'locks' the random numbers so your results are reproducible.
*   **Attributes**:
    *   `.ndim`: How many levels of brackets (dimensions) are there?
    *   `.shape`: What is the actual grid layout?
    *   `.size`: How many total items are inside?

```python
import numpy as np
# Setting a 'seed' ensures the 'random' numbers are the same every time we run this
np.random.seed(84)

x=np.random.randint(500,600,size=(3,3))
print(x)

# Checking the 'metadata' of our array
print(x.ndim)  # Number of dimensions (e.g., 2D)
print(x.size)  # Total number of elements inside
print(x.shape) # The exact grid size (rows, columns)

# Use .reshape() to change the look of an existing sequence
print(np.arange(10).reshape(2,5))
```

# ⚡ Vectorization (NumPy Math)

### Core Concepts:
*   **Broadcasting**: In NumPy, you don't need loops. If you do `array + 10`, it automatically adds 10 to every single element inside. This makes NumPy incredibly fast for data science.

```python
# Task 1: Basic range (start, stop, step)
a=np.arange(5,26,5)
print(a)

# Task 2: Getting specific points between two numbers
print(np.linspace(2,20,7)) # (start, stop, how_many_numbers)

# Task 3: Making a grid from a sequence
c=np.arange(12).reshape(3,4)

# Task 4 & 5: Zeros and Ones
print(np.zeros((2,5), dtype=int))
print(np.ones((3,3),dtype=int))

# Task 6: Identity matrix
print(np.identity(4))

# Task 7: Linear spacing practice
print(np.linspace(10,50,5))

# Task 8: Checking array attributes
print(c)
print(f"Shape: {c.shape}")

# Task 9: Math operations (NumPy does this to EVERY number in the array at once!)
print(c + 10)
print(c * 2)
print(c ** 2)
```
